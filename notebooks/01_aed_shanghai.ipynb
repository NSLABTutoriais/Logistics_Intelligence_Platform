## 03 — Localização e validação do dataset

### Objetivo

Localizar o arquivo original LaDe-P Shanghai e verificar sua existência
e seu tamanho antes de iniciar a leitura dos dados.

### Conceitos de Python

**Path:** classe da biblioteca `pathlib` utilizada para representar
e manipular caminhos de arquivos e diretórios.

**Caminho relativo:** localização definida a partir de um diretório
de referência, como a raiz do projeto.

**Caminho absoluto:** localização completa de um arquivo no sistema.

**exists():** método que verifica se um caminho existe.

**is_file():** método que verifica se o caminho representa um arquivo.

**stat():** método que retorna informações do arquivo, incluindo
seu tamanho em bytes.

### Boas práticas

- Preservar os dados originais.
- Evitar caminhos absolutos específicos de um computador.
- Validar a existência do arquivo antes de tentar carregá-lo.
- Separar a inspeção dos dados das transformações de ETL.

```python
from pathlib import Path

# Localização da raiz do projeto
projeto_dir = Path.cwd()

# Se o notebook estiver executando a partir de /notebooks,
# identificamos a pasta raiz do projeto.
if projeto_dir.name == "notebooks":
    projeto_dir = projeto_dir.parent

# Caminho do arquivo original
arquivo_parquet = (
    projeto_dir
    / "data"
    / "raw"
    / "pickup"
    / "pickup_sh-00000-of-00001-79fabe8088e723a2.parquet"
)

print("=== VALIDAÇÃO DO DATASET ===")
print(f"Raiz do projeto: {projeto_dir}")
print(f"Arquivo: {arquivo_parquet.name}")
print(f"Existe: {arquivo_parquet.exists()}")
print(f"É arquivo: {arquivo_parquet.is_file()}")

if arquivo_parquet.is_file():
    tamanho_bytes = arquivo_parquet.stat().st_size
    tamanho_mib = tamanho_bytes / (1024 ** 2)

    print(f"Tamanho (bytes): {tamanho_bytes:,}")
    print(f"Tamanho (MiB): {tamanho_mib:.2f}")
else:
    print("ATENÇÃO: arquivo Parquet não encontrado.")
```

### Código	Explicação
* Path.cwd()	Obtém o diretório atual de execução
* .name	Retorna o nome do último componente do caminho
* .parent	Obtém o diretório pai
* /	Operador de composição de caminhos da classe Path
* .exists()	Verifica se o caminho existe
* .is_file()	Confirma que é um arquivo
* .stat().st_size	Retorna o tamanho em bytes
*  1024 ** 2	Calcula 1.048.576 bytes, equivalentes a 1 MiB

## 04 — Inspeção dos metadados do Parquet

### Objetivo

Investigar a estrutura física do dataset antes de carregar os registros
em memória.

### Conceitos

**Metadados:** informações que descrevem um conjunto de dados,
como quantidade de registros, colunas e organização interna.

**Schema:** estrutura que define os nomes e tipos das colunas.

**Row group:** grupo de linhas armazenado internamente no Parquet.

**Compressão:** técnica utilizada para reduzir o espaço ocupado
pelos dados em disco.

### Por que esta inspeção é importante?

Permite compreender o volume e a estrutura dos dados antes de decidir
como realizar a leitura e o processamento.

Nesta etapa, utilizaremos PyArrow para consultar os metadados,
sem carregar o conteúdo completo do arquivo.

### 04.1 — Explicação do código Python

Nesta etapa, utilizamos a biblioteca **PyArrow** para consultar os
metadados do arquivo Parquet sem carregar todos os registros na memória.

#### 1. Importação da biblioteca

`import pyarrow.parquet as pq`

- `pyarrow`: biblioteca Python para processamento de dados colunares.
- `parquet`: módulo especializado na leitura e escrita de arquivos Parquet.
- `as pq`: cria um apelido para facilitar a utilização do módulo.

#### 2. Abertura do arquivo Parquet

`parquet_file = pq.ParquetFile(arquivo_parquet)`

- `pq.ParquetFile()`: cria um objeto para acessar o arquivo Parquet.
- `arquivo_parquet`: variável definida na etapa anterior, contendo
  o caminho do arquivo original.
- `parquet_file`: variável que armazena o objeto criado.

**Importante:** essa operação não carrega o dataset inteiro na memória.

#### 3. Consulta dos metadados

`metadata = parquet_file.metadata`

- `.metadata`: propriedade que retorna informações sobre a estrutura
  física do arquivo.
- `metadata`: variável utilizada para armazenar essas informações.

#### 4. Quantidade de registros

`metadata.num_rows`

Retorna a quantidade total de registros armazenados no arquivo Parquet.

#### 5. Quantidade de colunas

`metadata.num_columns`

Retorna o número de colunas existentes no dataset.

#### 6. Quantidade de row groups

`metadata.num_row_groups`

Retorna a quantidade de grupos de linhas utilizados na organização
interna do arquivo Parquet.

Os *row groups* permitem que mecanismos de processamento consultem
partes do arquivo sem necessariamente ler todos os registros.

#### 7. Versão do formato Parquet

`metadata.format_version`

Informa a versão do formato Parquet utilizada no arquivo.

#### 8. Ferramenta de criação

`metadata.created_by`

Identifica, quando disponível, a ferramenta ou biblioteca utilizada
para gerar o arquivo Parquet.

#### 9. Schema do dataset

`parquet_file.schema_arrow`

Retorna o esquema do arquivo no formato Apache Arrow, incluindo:

- Nome de cada coluna.
- Tipo físico ou lógico representado no Arrow.
- Informações adicionais do schema, quando disponíveis.

O schema permite identificar, por exemplo, se uma coluna está
armazenada como texto, número inteiro, número decimal ou timestamp.

**Atenção:** o tipo de uma coluna não garante seu significado
operacional. Essa interpretação será investigada posteriormente
na validação semântica.

#### 10. Formatação da saída

`print(f"Quantidade de registros: {metadata.num_rows:,}")`

Utilizamos uma **f-string**, identificada pelo prefixo `f`,
para inserir valores de variáveis diretamente em textos.

A expressão `{metadata.num_rows:,}` utiliza o especificador `:,`
para apresentar o número com separadores de milhares.

O separador utilizado pelo Python será uma vírgula, por exemplo:
`1,424,406`.

### Conclusão da etapa

A inspeção dos metadados permite conhecer a estrutura do dataset
antes de iniciar sua leitura completa.

Nesta etapa, não realizamos limpeza, transformação ou alteração
dos dados originais.

```python
import pyarrow.parquet as pq

# Abre o arquivo Parquet para inspeção de metadados
parquet_file = pq.ParquetFile(arquivo_parquet)

# Obtém os metadados
metadata = parquet_file.metadata

print("=== METADADOS DO PARQUET ===")

print(f"Quantidade de registros: {metadata.num_rows:,}")
print(f"Quantidade de colunas: {metadata.num_columns}")
print(f"Quantidade de row groups: {metadata.num_row_groups:,}")
print(f"Versão do formato Parquet: {metadata.format_version}")
print(f"Criado por: {metadata.created_by}")

print("\n=== SCHEMA DO DATASET ===")
print(parquet_file.schema_arrow)
```

### 05.4 — Validação semântica da coluna ds

#### Pergunta da investigação

A coluna `ds` representa consistentemente o mês e o dia
registrados em `pickup_time`?

#### Hipótese

Nos exemplos observados:

- `pickup_time = 07-08 10:10:00`
- `ds = 708`

Isso sugere que `ds` representa o mês e o dia no formato
numérico MDD ou MMDD, sem zeros à esquerda.

#### Metodologia

1. Ler somente as colunas necessárias.
2. Extrair mês e dia de `pickup_time`.
3. Calcular o valor esperado de `ds`.
4. Comparar o valor esperado com o valor armazenado.
5. Quantificar registros consistentes e divergentes.

#### Conceitos de Python e Pandas

- **Seleção de colunas:** permite carregar apenas os campos necessários.
- **Operações vetorizadas:** aplicam cálculos a uma coluna inteira.
- **Conversão de tipos:** transforma valores para permitir operações.
- **Máscara booleana:** identifica registros que atendem a uma condição.
- **Validação semântica:** verifica se os valores correspondem ao
  significado esperado de uma variável.

#### Cuidados

A análise não altera o arquivo original.

Uma divergência entre `ds` e `pickup_time` não significa
automaticamente que o registro esteja incorreto.

Precisaremos investigar possíveis diferenças de significado,
origem ou referência temporal.

```python
# Leitura apenas das colunas necessárias
df_ds = pd.read_parquet(
    arquivo_parquet,
    columns=["order_id", "ds", "pickup_time"]
)

# Extração do mês e do dia de pickup_time
partes_data = df_ds["pickup_time"].str.extract(
    r"^(\d{2})-(\d{2})\s"
)

# Conversão das partes extraídas para números
mes = pd.to_numeric(partes_data[0], errors="coerce")
dia = pd.to_numeric(partes_data[1], errors="coerce")

# Cálculo do valor esperado de ds
df_ds["ds_esperado"] = mes * 100 + dia

# Identificação de formatos temporais não reconhecidos
formato_invalido = df_ds["ds_esperado"].isna()

# Comparação entre ds e o valor esperado
df_ds["ds_consistente"] = (
    ~formato_invalido
    & df_ds["ds"].eq(df_ds["ds_esperado"])
)

# Resumo da validação
total = len(df_ds)
consistentes = int(df_ds["ds_consistente"].sum())
divergentes = total - consistentes
formatos_invalidos = int(formato_invalido.sum())

print("=== VALIDAÇÃO SEMÂNTICA DE DS ===")
print(f"Total de registros: {total:,}")
print(f"Registros consistentes: {consistentes:,}")
print(f"Registros divergentes: {divergentes:,}")
print(f"Formatos não reconhecidos: {formatos_invalidos:,}")
print(f"Percentual consistente: {consistentes / total:.2%}")

print("\n=== EXEMPLOS DE DIVERGÊNCIAS ===")

display(
    df_ds.loc[
        ~df_ds["ds_consistente"],
        ["order_id", "ds", "pickup_time", "ds_esperado"]
    ].head(10)
)
```

### 05.5 — Validação dos formatos temporais

#### Objetivo

Verificar se os campos temporais do LaDe-P Shanghai apresentam
formatos consistentes e identificar possíveis valores inválidos.

#### Colunas investigadas

- `accept_time`: momento do aceite da coleta.
- `time_window_start`: início da janela de coleta.
- `time_window_end`: fim da janela de coleta.
- `pickup_time`: momento registrado da coleta.
- `accept_gps_time`: horário do registro GPS associado ao aceite.
- `pickup_gps_time`: horário do registro GPS associado à coleta.

#### Hipótese

Os campos temporais seguem o formato:

`MM-DD HH:MM:SS`

Exemplo: `07-08 10:10:00`

#### Metodologia

1. Ler somente as seis colunas temporais.
2. Identificar valores ausentes.
3. Verificar a estrutura textual dos horários.
4. Identificar formatos não reconhecidos.
5. Apresentar exemplos de valores inválidos.

#### Conceitos

**Expressão regular (regex):**
padrão utilizado para verificar a estrutura de textos.

**Valor ausente:**
informação não preenchida, representada no Pandas
por valores como `None`, `NaN` ou `NaT`.

**Validação sintática:**
verifica se um valor segue um formato esperado.

**Validação semântica:**
verifica se o valor possui significado válido.

Por exemplo, `13-45 10:00:00` segue a estrutura textual
MM-DD HH:MM:SS, mas não representa uma data válida.

#### Cuidados

A validação do formato não comprova a ordem cronológica
dos eventos nem a validade completa das datas.

Essas verificações serão realizadas separadamente.

```python
# Colunas temporais que serão investigadas
colunas_temporais = [
    "accept_time",
    "time_window_start",
    "time_window_end",
    "pickup_time",
    "accept_gps_time",
    "pickup_gps_time",
]

# Leitura seletiva do Parquet
df_temporal = pd.read_parquet(
    arquivo_parquet,
    columns=colunas_temporais
)

# Expressão regular para o formato MM-DD HH:MM:SS
padrao_temporal = (
    r"\d{2}-\d{2} "
    r"\d{2}:\d{2}:\d{2}"
)

resultados_temporais = []

for coluna in colunas_temporais:

    serie = df_temporal[coluna]

    # Identificação de valores ausentes
    ausentes = serie.isna()

    # Validação da estrutura textual
    formato_valido = serie.str.fullmatch(
        padrao_temporal,
        na=False
    )

    # Valores preenchidos que não seguem o padrão
    formato_invalido = ~ausentes & ~formato_valido

    resultados_temporais.append({
        "coluna": coluna,
        "total": len(serie),
        "ausentes": int(ausentes.sum()),
        "formato_valido": int(formato_valido.sum()),
        "formato_invalido": int(formato_invalido.sum())
    })

# Organização dos resultados em DataFrame
df_resultado_temporal = pd.DataFrame(
    resultados_temporais
)

print("=== VALIDAÇÃO DOS FORMATOS TEMPORAIS ===")

display(df_resultado_temporal)

print("\n=== EXEMPLOS DE FORMATOS INVÁLIDOS ===")

for coluna in colunas_temporais:

    serie = df_temporal[coluna]

    mascara_invalida = (
        serie.notna()
        & ~serie.str.fullmatch(padrao_temporal, na=False)
    )

    if mascara_invalida.any():
        print(f"\nColuna: {coluna}")
        display(serie[mascara_invalida].head(5))
```

#### Metodologia

1. Reutilizar os campos temporais da etapa anterior.

3. Validar a conversão para datetime.
4. Identificar valores fora do período documentado.

#### Conceitos

**Validação documental:**

```python

# Etapa 05.6 — Validação do calendário e período

# Período informado na documentação oficial do LaDe
inicio_documentado = pd.Timestamp("2022-05-01")
fim_documentado = pd.Timestamp("2022-10-31 23:59:59")

# Ano identificado na documentação
ano_documentado = 2022

resultados_periodo = []
exemplos_problemas = {}

for coluna in colunas_temporais:

    # Reutilizar os dados da Etapa 05.5
    serie = df_temporal[coluna]

    # Considerar somente os valores preenchidos
    preenchidos = serie.dropna()

    # Reconstruir a data com o ano documentado
    textos_com_ano = (
        str(ano_documentado) + "-" + preenchidos
    )

    # Converter para datetime
    datas = pd.to_datetime(
        textos_com_ano,
        format="%Y-%m-%d %H:%M:%S",
        errors="coerce",
        exact=True
    )

    # Identificar datas impossíveis ou formatos inválidos
    mascara_invalida = datas.isna()

    # Identificar datas válidas fora do período documentado
    mascara_fora_periodo = (
        datas.notna()
        & ~datas.between(inicio_documentado, fim_documentado)
    )

    # Guardar indicadores
    resultados_periodo.append({
        "coluna": coluna,
        "preenchidos": len(preenchidos),
        "datas_invalidas": int(mascara_invalida.sum()),
        "fora_periodo": int(mascara_fora_periodo.sum()),
        "primeira_data": datas.min(),
        "ultima_data": datas.max()
    })

    # Guardar exemplos para investigação
    mascara_problema = mascara_invalida | mascara_fora_periodo

    exemplos_problemas[coluna] = preenchidos[
        mascara_problema
    ].head(5)

# Consolidar resultados
df_resultado_periodo = pd.DataFrame(resultados_periodo)

print("=== VALIDAÇÃO DE CALENDÁRIO E PERÍODO ===")
print(f"Período documentado: {inicio_documentado} a {fim_documentado}")

display(df_resultado_periodo)

print("\n=== EXEMPLOS DE INCONSISTÊNCIAS ===")

for coluna, exemplos in exemplos_problemas.items():
    if not exemplos.empty:
        print(f"\nColuna: {coluna}")
        display(exemplos)
```

### 05.7 — Validação cronológica dos eventos

#### Objetivo

Investigar a consistência cronológica entre os eventos de aceite, a janela programada e a coleta efetiva.

Essa validação é necessária antes de definir a variável-alvo dos modelos de classificação e regressão.

#### Perguntas da investigação

1. Existem coletas registradas antes do aceite?
2. Existem janelas programadas cujo início é posterior ao término?
3. Existem coletas realizadas antes do início da janela?
4. Existem coletas realizadas depois do término da janela?

#### Campos utilizados

- `accept_time`: momento do aceite.
- `time_window_start`: início da janela programada.
- `time_window_end`: término da janela programada.
- `pickup_time`: momento da coleta efetiva.

#### Referência documental

Utilizaremos o ano de 2022, conforme o período documentado do LaDe.

Fonte: https://github.com/wenhaomin/LaDe/blob/master/readme.md

#### Metodologia

1. Reutilizar os campos temporais já carregados.
2. Reconstruir as datas com o ano documentado.
3. Comparar os horários dos eventos.
4. Contabilizar as ocorrências de cada situação.
5. Interpretar os resultados considerando a operação logística.

#### Cuidados

Uma coleta realizada antes do início da janela não será automaticamente considerada um erro.

Uma coleta posterior ao término da janela será inicialmente tratada como candidata a atraso.

A definição final de atraso dependerá da interpretação operacional das janelas de coleta.

Nenhum registro será excluído ou alterado nesta etapa.

```python

# Etapa 05.7 — Validação cronológica dos eventos

# Campos necessários para a investigação
colunas_eventos = [
    "accept_time",
    "time_window_start",
    "time_window_end",
    "pickup_time"
]

# Criar DataFrame de trabalho sem modificar o original
df_cronologia = df_temporal[colunas_eventos].copy()

# Reconstruir datas utilizando o ano documentado
for coluna in colunas_eventos:

    df_cronologia[coluna] = pd.to_datetime(
        "2022-" + df_cronologia[coluna],
        format="%Y-%m-%d %H:%M:%S",
        errors="coerce"
    )

# Identificar situações cronológicas
coleta_antes_aceite = (
    df_cronologia["pickup_time"]
    < df_cronologia["accept_time"]
)

janela_invertida = (
    df_cronologia["time_window_start"]
    > df_cronologia["time_window_end"]
)

coleta_antes_janela = (
    df_cronologia["pickup_time"]
    < df_cronologia["time_window_start"]
)

coleta_apos_janela = (
    df_cronologia["pickup_time"]
    > df_cronologia["time_window_end"]
)

# Consolidar resultados
resultados_cronologia = {
    "Coleta antes do aceite": int(coleta_antes_aceite.sum()),
    "Janela invertida": int(janela_invertida.sum()),
    "Coleta antes da janela": int(coleta_antes_janela.sum()),
    "Coleta após a janela": int(coleta_apos_janela.sum())
}

df_resultado_cronologia = pd.DataFrame(
    resultados_cronologia.items(),
    columns=["situacao", "quantidade"]
)

df_resultado_cronologia["percentual"] = (
    df_resultado_cronologia["quantidade"]
    / len(df_cronologia)
    * 100
)

print("=== VALIDAÇÃO CRONOLÓGICA ===")
print(f"Total de registros: {len(df_cronologia):,}")

display(df_resultado_cronologia)
```

### 05.8 — Análise da duração e gravidade dos atrasos

#### 05.8.1 — Análise estatística da duração dos atrasos

#### Objetivo

Calcular a duração dos atrasos nas coletas e investigar sua distribuição, utilizando o término da janela programada como referência.

#### Referência documental

Segundo a documentação do LaDe-P:

- `pickup_time`: horário efetivo da coleta.
- `time_window_end`: término da janela programada de coleta.

Fonte: https://github.com/wenhaomin/LaDe/blob/master/readme.md

#### Definição analítica provisória

Consideraremos atraso quando:

`pickup_time > time_window_end`

A duração do atraso será calculada em minutos:

`atraso_minutos = pickup_time - time_window_end`

Para coletas realizadas dentro do prazo ou antecipadamente, o atraso será igual a zero.

Essa definição é uma regra analítica do laboratório, não uma confirmação de SLA contratual.

#### Perguntas da investigação

1. Quantas coletas apresentam atraso?
2. Qual é o atraso médio?
3. Qual é o atraso mediano?
4. Qual é o maior atraso observado?
5. Como os atrasos estão distribuídos?
6. Existem valores extremos que merecem investigação?

#### Conceitos estatísticos

**Média:** soma dos valores dividida pela quantidade de observações.

**Mediana:** valor central de uma distribuição ordenada.

**Percentil:** valor abaixo do qual se encontra determinada proporção das observações.

**Outlier:** observação muito distante do comportamento predominante.

#### Cuidados metodológicos

A média será calculada somente entre as coletas atrasadas, evitando que os registros sem atraso distorçam a interpretação da gravidade.

Os valores extremos serão inicialmente preservados.

Nenhum registro será excluído ou corrigido nesta etapa.

Os resultados poderão orientar a construção da variável-alvo de regressão e a definição de faixas de gravidade.

```python

# Etapa 05.8.1 — Duração e gravidade dos atrasos

# Criar DataFrame de análise
df_atrasos = df_cronologia.copy()

# Calcular a diferença em minutos
df_atrasos["diferenca_minutos"] = (
    (
        df_atrasos["pickup_time"]
        - df_atrasos["time_window_end"]
    )
    .dt.total_seconds() / 60
)

# Definir atraso: somente diferenças positivas
df_atrasos["atraso_minutos"] = (
    df_atrasos["diferenca_minutos"].clip(lower=0)
)

# Selecionar somente as coletas atrasadas
atrasos_positivos = df_atrasos.loc[
    df_atrasos["atraso_minutos"] > 0,
    "atraso_minutos"
]

# Calcular indicadores estatísticos
resumo_atrasos = pd.Series({
    "Total de coletas": len(df_atrasos),
    "Coletas atrasadas": len(atrasos_positivos),
    "Percentual de atraso (%)": (
        len(atrasos_positivos) / len(df_atrasos) * 100
    ),
    "Atraso médio (min)": atrasos_positivos.mean(),
    "Atraso mediano (min)": atrasos_positivos.median(),
    "Percentil 75 (min)": atrasos_positivos.quantile(0.75),
    "Percentil 90 (min)": atrasos_positivos.quantile(0.90),
    "Percentil 95 (min)": atrasos_positivos.quantile(0.95),
    "Percentil 99 (min)": atrasos_positivos.quantile(0.99),
    "Maior atraso (min)": atrasos_positivos.max()
})

print("=== ANÁLISE DA DURAÇÃO DOS ATRASOS ===")

display(
    resumo_atrasos.to_frame(name="valor")
)

print("\n=== DISTRIBUIÇÃO DOS ATRASOS POSITIVOS ===")

display(
    atrasos_positivos.describe()
)
```

A diferença original permanece disponível na coluna `diferenca_minutos`.
