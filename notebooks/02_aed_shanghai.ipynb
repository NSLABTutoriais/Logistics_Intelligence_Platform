#### Onde entra o RAG?

O **RAG (Retrieval-Augmented Generation)** permite combinar modelos de linguagem com informações recuperadas de fontes de conhecimento.

É importante distinguir suas responsabilidades:

- **Machine Learning:** identifica padrões e produz previsões.
- **RAG:** recupera informações documentadas para fundamentar respostas.
- **LLM:** interpreta perguntas e gera respostas em linguagem natural.
- **Agentes:** utilizam ferramentas autorizadas para consultar dados, executar análises e acessar modelos.

O RAG não substitui os modelos de classificação ou regressão.

A integração dessas tecnologias poderá permitir que a aplicação explique indicadores, apresente evidências e apoie a investigação de situações operacionais.

## 00 — Pré-verificação do ambiente e das dependências

**Objetivo:** verificar, antes dos imports da AED, o interpretador Python, as bibliotecas obrigatórias, as versões instaladas e a correspondência com as versões declaradas em `requirements.txt` (quando presente). Esta verificação **não instala nem altera** pacotes. Se faltar uma dependência ou houver divergência, o notebook interrompe a execução com diagnóstico explícito. A comparação considera linhas simples no formato `pacote==versão`; outras especificações são sinalizadas para revisão manual.

```python
from pathlib import Path
import sys
from importlib import metadata, util

# Nome da distribuição (pip) -> módulo importável (Python).
REQUIRED = {
    "numpy": "numpy", "pandas": "pandas", "pyarrow": "pyarrow",
    "matplotlib": "matplotlib", "ipython": "IPython",
}

print("Interpretador:", sys.executable)
print("Versão Python:", sys.version.split()[0])
if sys.version_info < (3, 11):
    raise RuntimeError("É necessário Python 3.11 ou superior.")

missing = []
installed = {}
for distribution, module in REQUIRED.items():
    try:
        version = metadata.version(distribution)
        installed[distribution] = version
        available = util.find_spec(module) is not None
        print(f"{distribution}: {version} | módulo importável: {available}")
        if not available:
            missing.append(distribution)
    except metadata.PackageNotFoundError:
        missing.append(distribution)
        print(f"{distribution}: NÃO INSTALADO")

root = Path.cwd()
if not (root / "requirements.txt").exists() and (root.parent / "requirements.txt").exists():
    root = root.parent
requirements = root / "requirements.txt"
issues = []
if requirements.exists():
    print("Arquivo de dependências:", requirements)
    from packaging.requirements import Requirement
    from packaging.utils import canonicalize_name
    for line in requirements.read_text(encoding="utf-8-sig").splitlines():
        line = line.strip()
        if not line or line.startswith("#") or line.startswith(("-r", "--")):
            continue
        try:
            req = Requirement(line)
            name = canonicalize_name(req.name)
            current = metadata.version(req.name)
            if req.specifier and current not in req.specifier:
                issues.append(f"{req.name}: instalado {current}; esperado {req.specifier}")
        except metadata.PackageNotFoundError:
            issues.append(f"{line}: não instalado")
        except Exception as exc:
            issues.append(f"{line}: não foi possível verificar ({exc})")
else:
    print("requirements.txt não encontrado; comparar versões manualmente.")

if missing or issues:
    raise RuntimeError("Dependências pendentes: " + "; ".join(missing + issues))
print("Pré-verificação concluída: dependências obrigatórias disponíveis e requisitos verificados.")
```

### Explicação do código — Pré-verificação

#### 1. Interpretador e versão do Python
- `sys.executable` identifica **qual Python executa o kernel**; o terminal do VS Code pode apontar para outro interpretador.
- `sys.version_info < (3, 11)` aplica uma condição de versão mínima. `raise RuntimeError(...)` interrompe a análise de maneira explícita.

#### 2. Dependências obrigatórias
- `REQUIRED` é um **dicionário** que relaciona o nome do pacote instalado via `pip` ao módulo importado pelo Python; por exemplo, `ipython` → `IPython`.
- `for distribution, module in REQUIRED.items()` percorre os pares do dicionário, sem repetir código.
- `metadata.version()` consulta a versão da distribuição instalada; `util.find_spec()` verifica se o módulo é localizável, **sem executar sua importação**.
- `try` / `except metadata.PackageNotFoundError` trata pacotes ausentes sem interromper a listagem das demais dependências.

#### 3. Comparação com o arquivo de requisitos
- `Path.cwd()` obtém a pasta de execução; `.parent` permite localizar a raiz quando o kernel inicia em `notebooks/`.
- `read_text(encoding="utf-8-sig")` lê o arquivo e aceita um possível marcador BOM UTF-8.
- `Requirement(line)` interpreta especificações de versões; `canonicalize_name()` normaliza nomes de pacotes.
- `current not in req.specifier` identifica versões que não atendem ao intervalo exigido.

> **Limitação:** a checagem não instala bibliotecas, não testa a importação real de todos os módulos e não substitui `python -m pip check`. Linhas especiais de requisitos podem exigir revisão manual.

**Critério de aprovação:** nenhuma dependência obrigatória ausente e nenhuma incompatibilidade detectada no arquivo de requisitos.

## 03 — Inventário estrutural das 19 colunas

Comparar tipos, valores de exemplo, cardinalidade e memória. Identificadores podem parecer numéricos, mas não devem ser tratados automaticamente como medidas quantitativas.

```python
inventory = pd.DataFrame({
    "coluna": df.columns,
    "dtype": [str(df[c].dtype) for c in df.columns],
    "na": [int(df[c].isna().sum()) for c in df.columns],
    "na_pct": [safe_pct(df[c].isna().sum(), len(df)) for c in df.columns],
    "distintos": [int(df[c].nunique(dropna=True)) for c in df.columns],
    "exemplos": [repr(df[c].dropna().iloc[:3].tolist())[:110] for c in df.columns],
    "mem_mib": [round(df[c].memory_usage(deep=True)/2**20, 2) for c in df.columns]
})
show("Inventário completo", inventory, 30)
print("Memória total MiB:", round(df.memory_usage(deep=True).sum()/2**20, 2))
```

### Explicação do código — 03 — Inventário estrutural das 19 colunas

#### 1. Inventário com compreensões de listas
- `pd.DataFrame({...})` cria uma tabela-resumo: **cada linha representa uma coluna da fonte**.
- `[... for c in df.columns]` é uma *list comprehension*: aplica a mesma verificação às 19 variáveis.
- `df[c].dtype` informa o tipo inferido; `str(...)` o torna legível na tabela.

#### 2. Qualidade, cardinalidade e exemplos
- `.isna().sum()` conta ausências; `safe_pct(..., len(df))` converte o total em percentual.
- `.nunique(dropna=True)` conta valores distintos **sem considerar nulos**: útil para detectar identificadores, categorias e constantes.
- `.dropna().iloc[:3].tolist()` captura até três exemplos não nulos; `repr(...)[:110]` limita a largura visual, sem alterar os dados.

#### 3. Memória
- `.memory_usage(deep=True)` estima consumo de memória, incluindo objetos Python quando aplicável.
- `2**20` converte bytes para MiB (1 MiB = 1.048.576 bytes).

> **Atenção:** alta cardinalidade não é defeito por si só; pode ser esperada em IDs. O tipo inferido também não substitui a validação do domínio de negócio.

## 04 — Nulos, duplicatas e integridade de identificadores

Medir ausências, padrões conjuntos de GPS, linhas idênticas, repetição de `order_id` e combinações de chaves candidatas. **Repetição não equivale automaticamente a erro:** investigar granularidade e documentação.

```python
show("Nulos por coluna", inventory[["coluna", "na", "na_pct"]].sort_values("na_pct", ascending=False), 30)
print("Linhas integralmente duplicadas:", int(df.duplicated().sum()))
for keys in [["order_id"], ["order_id", "ds"], ["order_id", "courier_id"]]:
    print("Duplicatas na chave", keys, ":", int(df.duplicated(subset=keys).sum()))
for t, x, y in [("accept_gps_time", "accept_gps_lng", "accept_gps_lat"), ("pickup_gps_time", "pickup_gps_lng", "pickup_gps_lat")]:
    miss = df[[t, x, y]].isna()
    show(f"Padrões de ausência {t}", miss.value_counts(dropna=False).rename("linhas").reset_index(), 10)
print("Colunas constantes (<=1 valor não nulo):", inventory.loc[inventory.distintos <= 1, "coluna"].tolist())
print("Linhas com qualquer nulo:", int(df.isna().any(axis=1).sum()))
```

### Explicação do código — 04 — Nulos, duplicatas e integridade de identificadores

#### 1. Ausências e ordenação
- `inventory[[...]]` seleciona colunas do inventário; `.sort_values("na_pct", ascending=False)` prioriza campos com maior proporção de nulos.
- `.isna().any(axis=1)` marca linhas com **ao menos um campo ausente**.

#### 2. Duplicatas: conceitos distintos
- `df.duplicated()` identifica linhas integralmente repetidas, mantendo a primeira ocorrência como referência.
- `df.duplicated(subset=keys)` testa diferentes **chaves candidatas**: `order_id`, `order_id + ds` e `order_id + courier_id`.
- Uma duplicata por chave não prova erro: é preciso confirmar a **granularidade** do dataset e a regra de unicidade.

#### 3. Ausência conjunta de GPS
- `df[[t, x, y]].isna()` constrói uma tabela booleana de ausência de tempo, longitude e latitude.
- `.value_counts(dropna=False)` conta combinações de presença/ausência; `.reset_index()` transforma o resultado em tabela legível.
- `inventory.distintos <= 1` sinaliza colunas constantes ou sem valores não nulos.

**Interpretação futura:** distinguir nulos estruturais (evento não registrado) de falhas de qualidade; nenhuma linha é removida nesta etapa.

## 09 — Dimensão geográfica e GPS

Verificar latitude/longitude, completude de pares, dispersão, coordenadas fora de limites globais e diferenças entre pontos de referência e GPS. Distâncias são estimadas em linha reta (Haversine), **não representam rotas reais**. Não presumir que `lng/lat` sejam origem ou destino sem documentação.


**Mapa exploratório:** representar graficamente os três pares de coordenadas em longitude/latitude, com amostragem reprodutível. A visualização não é um mapa viário nem uma rota logística.

```python
for lon, lat in COORD_PAIRS:
    x = pd.to_numeric(df[lon], errors="coerce"); y = pd.to_numeric(df[lat], errors="coerce")
    complete = x.notna() & y.notna()
    invalid = complete & (~x.between(-180,180) | ~y.between(-90,90))
    partial = x.isna() ^ y.isna()
    print(f"{lon}/{lat}: completos={complete.sum():,}, parciais={partial.sum():,}, fora dos limites globais={invalid.sum():,}")
    print("  lng min/max:", x.min(), x.max(), "lat min/max:", y.min(), y.max())

def haversine_km(lon1, lat1, lon2, lat2):
    r = 6371.0088
    p1, p2 = np.radians(lat1), np.radians(lat2)
    dp = p2-p1; dl = np.radians(lon2-lon1)
    a = np.sin(dp/2)**2 + np.cos(p1)*np.cos(p2)*np.sin(dl/2)**2
    return 2*r*np.arcsin(np.sqrt(np.clip(a, 0, 1)))

for prefix in ["pickup_gps", "accept_gps"]:
    distance = haversine_km(pd.to_numeric(df["lng"], errors="coerce"), pd.to_numeric(df["lat"], errors="coerce"), pd.to_numeric(df[f"{prefix}_lng"], errors="coerce"), pd.to_numeric(df[f"{prefix}_lat"], errors="coerce"))
    analysis[f"dist_{prefix}_referencia_km"] = distance
    print(prefix, "distância em linha reta (km):")
    display(distance.describe(percentiles=[.5,.9,.95,.99]))
show("Contagem por região", df.groupby("region_id", dropna=False).size().sort_values(ascending=False).rename("coletas").reset_index(), 20)
show("Contagem por AOI", df.groupby("aoi_id", dropna=False).size().sort_values(ascending=False).rename("coletas").reset_index(), 20)


# Visualização geográfica: zoom de Shanghai + panorama de coordenadas afastadas.
# Não é um mapa de ruas; usa longitude e latitude reais, sem rotas inferidas.
MAP_SAMPLE_PER_LAYER = 20000
map_layers = [
    ("lng", "lat", "Referência do pedido"),
    ("accept_gps_lng", "accept_gps_lat", "GPS no aceite"),
    ("pickup_gps_lng", "pickup_gps_lat", "GPS na coleta"),
]
coords_by_layer = {}
for lon_col, lat_col, label in map_layers:
    coords = pd.DataFrame({
        "lon": pd.to_numeric(df[lon_col], errors="coerce"),
        "lat": pd.to_numeric(df[lat_col], errors="coerce"),
    }).replace([np.inf, -np.inf], np.nan).dropna()
    coords = coords.loc[coords.lon.between(-180, 180) & coords.lat.between(-90, 90)]
    coords_by_layer[label] = coords

# Os limites locais são definidos pela referência do pedido, com pequena margem.
reference = coords_by_layer["Referência do pedido"]
if not reference.empty:
    lon_min, lon_max = reference.lon.min(), reference.lon.max()
    lat_min, lat_max = reference.lat.min(), reference.lat.max()
    lon_pad = max((lon_max - lon_min) * .08, .02)
    lat_pad = max((lat_max - lat_min) * .08, .02)
    bounds = (lon_min-lon_pad, lon_max+lon_pad, lat_min-lat_pad, lat_max+lat_pad)
    fig, axes = plt.subplots(1, 2, figsize=(18, 8))
    for label, coords in coords_by_layer.items():
        if coords.empty:
            continue
        sampled = coords.sample(n=min(MAP_SAMPLE_PER_LAYER, len(coords)), random_state=42)
        inside_all = coords.lon.between(bounds[0], bounds[1]) & coords.lat.between(bounds[2], bounds[3])
        print(f"{label}: {len(coords):,} coordenadas válidas; {int(inside_all.sum()):,} dentro da área de referência; {int((~inside_all).sum()):,} fora")
        axes[0].scatter(sampled.lon, sampled.lat, s=2, alpha=.20, rasterized=True, label=label)
        axes[1].scatter(sampled.lon, sampled.lat, s=2, alpha=.20, rasterized=True, label=label)
    axes[0].set_xlim(bounds[0], bounds[1]); axes[0].set_ylim(bounds[2], bounds[3])
    axes[0].set_title("Shanghai — zoom na área dos pedidos")
    axes[1].set_title("Panorama geral — inclusive coordenadas afastadas")
    for ax in axes:
        ax.set_xlabel("Longitude (°)"); ax.set_ylabel("Latitude (°)")
        ax.grid(alpha=.2)
        ax.legend(markerscale=5)
        # Corrige a distorção aproximada de longitude em relação à latitude.
        ax.set_aspect(1 / np.cos(np.deg2rad((lat_min + lat_max) / 2)), adjustable="box")
    plt.tight_layout(); plt.show()
else:
    print("Não foi possível definir o enquadramento: referência sem coordenadas válidas.")

# Visão de densidade: evita que milhares de pontos sobrepostos escondam concentrações.
if not reference.empty:
    fig, ax = plt.subplots(figsize=(11, 8))
    density = ax.hexbin(reference.lon, reference.lat, gridsize=85, mincnt=1, bins="log")
    ax.set_xlim(bounds[0], bounds[1]); ax.set_ylim(bounds[2], bounds[3])
    ax.set_title("Shanghai — densidade de referências de pedidos (escala logarítmica)")
    ax.set_xlabel("Longitude (°)"); ax.set_ylabel("Latitude (°)")
    ax.set_aspect(1 / np.cos(np.deg2rad((lat_min + lat_max) / 2)), adjustable="box")
    fig.colorbar(density, ax=ax, label="Contagem por hexágono (escala log)")
    plt.tight_layout(); plt.show()
```

### Explicação do código — 09 — Dimensão geográfica e GPS

#### 1. Qualidade e significado das coordenadas
- `lng`/`lat` representam a referência geográfica do pedido; `accept_gps_*` e `pickup_gps_*` registram posições GPS em eventos diferentes.
- A checagem de pares completos e limites globais é apenas uma primeira validação. Um ponto pode estar dentro dos limites mundiais e ainda ser incompatível com a operação em Shanghai.
- `haversine_km()` calcula a distância aproximada em linha reta sobre a superfície terrestre. **Não** representa percurso, tempo de deslocamento ou rota.

#### 2. Por que o mapa anterior parecia vazio?
- Alguns pontos GPS estão muito afastados da região dos pedidos. Ao ajustar automaticamente os eixos para incluir todos, a área principal fica visualmente comprimida.

## 10 — Padrões operacionais, horários e associações

Segmentar por região, tipo de AOI, entregador e hora de aceite, sem usar previsão temporal. A taxa provisória de atraso é apenas descritiva. Correlação é associação, não causalidade; observar tamanhos de grupo.

```python
analysis["hora_aceite"] = parsed["accept_time"].dt.hour
analysis["hora_coleta"] = parsed["pickup_time"].dt.hour
for group in ["region_id", "aoi_type", "courier_id"]:
    agg = pd.DataFrame({group: df[group], "atrasada": analysis["atrasada_provisoria"], "atraso_min": analysis["atraso_min_provisorio"]}).groupby(group, dropna=False).agg(
        registros=("atrasada", "size"), elegiveis=("atrasada", "count"), taxa_atraso=("atrasada", "mean"), atraso_medio_min=("atraso_min", "mean")
    ).sort_values("registros", ascending=False)
    show(f"Indicadores por {group} (taxa como proporção)", agg.reset_index(), 15)
show("Por hora de aceite", pd.DataFrame({"hora":analysis["hora_aceite"], "atrasada":analysis["atrasada_provisoria"]}).groupby("hora", dropna=False).agg(registros=("atrasada", "size"), taxa_atraso=("atrasada", "mean")).reset_index(), 30)
assoc = analysis.select_dtypes(include="number").corr(method="spearman", min_periods=100)
show("Matriz de correlação de Spearman (métricas derivadas)", assoc, 30)
print("Nota: métricas construídas com os mesmos timestamps têm correlações mecânicas; não interpretá-las como descoberta causal.")
```

### Explicação do código — 10 — Padrões operacionais, horários e associações

#### 1. Componentes horários
- `.dt.hour` extrai a hora (0–23) de séries datetime para estudar distribuição operacional, **sem forecasting**.

#### 2. Agregações por segmento
- `pd.DataFrame({...})` reúne grupo, alvo provisório e atraso em uma estrutura auxiliar.
- `.groupby(group, dropna=False)` separa registros por região, tipo de AOI ou entregador, incluindo grupos ausentes.
- `.agg(registros=(..., "size"), elegiveis=(..., "count"), ...)` aplica **agregações nomeadas**:
  - `size`: todas as linhas do grupo;
  - `count`: somente valores não nulos;
  - `mean` do indicador booleano: taxa entre observações elegíveis;
  - `mean` do atraso: média dos minutos provisórios.
- `.reset_index()` transforma o agrupamento em tabela convencional.

#### 3. Associação entre métricas
- `.select_dtypes(include="number")` seleciona colunas numéricas auxiliares.
- `.corr(method="spearman", min_periods=100)` calcula associação **monotônica baseada em postos** em pares com observações suficientes.

> **Cautela:** associação não demonstra causalidade; métricas derivadas dos mesmos timestamps podem ter correlações mecânicas. Segmentos com poucos registros exigem atenção à estabilidade das taxas.
